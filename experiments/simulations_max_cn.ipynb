## Simulations for 1 sample (max_cn = 4)

```python
# set up parameters
n_trials = 5
num_mutations = 20
num_cells = 3000
num_CNAs = 3
num_nodes = 6 
num_samples= 1
num_regions = 20
save_path = os.path.join(os.getcwd(), "simulations_max_cn_4", "sims_1samples")
max_cn=4
seed = 636546773 # randomly generated seed 
```

```python
results1_df = run_simulations(n_trials,
                             num_mutations,
                             num_cells, 
                             num_CNAs,
                             num_nodes,
                             num_samples,
                             num_regions=num_regions,
                             max_cn=max_cn,
                             save_path=save_path,
                             include_LoPhy=True,
                             include_COMPASS=False,
                             include_SCITE=False,
                             include_LACE=False,
                             seed=seed,
                             plot_tree=True,
                             LoPhy_kwargs={"hom_precision":15.0, "het_precision":4.0, "fp":0.02, "fn":0.05 , "max_cn":max_cn})
```

```python
results1_df.to_csv(os.path.join(save_path, "sim_evaluation.csv"))
```

```python
results_dfimport matplotlib.pyplot as plt
import seaborn as sns

# Set up the 2x2 subplot grid
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# List of metrics and corresponding subplot positions
metrics = [("MCN-MAE", "MCN-MAE"),
           ("TCN_MAE", "TCN_MAE"),
           ("FER", "FER"),
           ("Run time", "Run time")]

# Flatten axes for easy iteration
axes = axes.flatten()

# Plot each metric
for ax, (metric_col, title) in zip(axes, metrics):
    sns.boxplot(data=results_df, x="Samples", y=metric_col, hue="Model", palette="deep", ax=ax, hue_order=["LoPhy", "COMPASS", "SCITE", "LACE"])
    ax.set_title(title)
    ax.set_ylabel(title)
    ax.set_xlabel("Samples")
    ax.legend_.remove()  # Remove individual legends

# Add a single shared legend outside the top-right
handles, labels = axes[0].get_legend_handles_labels()
fig.legend(handles, labels, loc='upper center', bbox_to_anchor=(0.5, 1.05), ncol=len(labels), frameon=False)

plt.tight_layout(rect=[0, 0, 1, 0.98])  # Leave space for legend
plt.show()
```
