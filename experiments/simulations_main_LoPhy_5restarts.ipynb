```python
df1 = pd.read_csv(os.path.join(os.getcwd(), "simulations_main_5restarts", "sims_1samples", "sim_evaluation.csv"), header=0, index_col=0)
df2 = pd.read_csv(os.path.join(os.getcwd(), "simulations_main_5restarts", "sims_2samples", "sim_evaluation.csv"), header=0, index_col=0)
df3 = pd.read_csv(os.path.join(os.getcwd(), "simulations_main_5restarts", "sims_3samples", "sim_evaluation.csv"), header=0, index_col=0)
df4 = pd.read_csv(os.path.join(os.getcwd(), "simulations_main_5restarts", "sims_4samples", "sim_evaluation.csv"), header=0, index_col=0)
df5 = pd.read_csv(os.path.join(os.getcwd(), "simulations_main_5restarts", "sims_5samples", "sim_evaluation.csv"), header=0, index_col=0)

df1["Samples"] = 1
df2["Samples"] = 2
df3["Samples"] = 3
df4["Samples"] = 4
df5["Samples"] = 5
```

```python
results_df = pd.concat([df2,df3,df4,df5])
results_df = results_df.reset_index(drop=True)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set up the 2x2 subplot grid
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# List of metrics and corresponding subplot positions
metrics = [("MCN-MAE", "MCN-MAE"),
           ("TCN_MAE", "TCN_MAE"),
           ("FER", "FER"),
           ("Run time", "Run time")]

# Flatten axes for easy iteration
axes = axes.flatten()

# Plot each metric
for ax, (metric_col, title) in zip(axes, metrics):
    sns.boxplot(data=results_df, x="Samples", y=metric_col, hue="Model", palette="deep", ax=ax, hue_order=["LoPhy", "COMPASS", "SCITE", "LACE"])
    ax.set_title(title)
    ax.set_ylabel(title)
    ax.set_xlabel("Samples")
    ax.legend_.remove()  # Remove individual legends

# Add a single shared legend outside the top-right
handles, labels = axes[0].get_legend_handles_labels()
fig.legend(handles, labels, loc='upper center', bbox_to_anchor=(0.5, 1.05), ncol=len(labels), frameon=False)

plt.tight_layout(rect=[0, 0, 1, 0.98])  # Leave space for legend
plt.show()
```
