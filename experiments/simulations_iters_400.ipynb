```python
import os, sys
from simulation_utils import run_simulations, load_tree
import oncophylo as op
from anndata import io
import networkx as nx
```

## Simulations for 1 sample

```python
# set up parameters
n_trials = 10
num_mutations = 20
num_cells = 3000
num_CNAs = 3
num_nodes = 6 
num_samples= 1
num_regions = 20
iters=400
save_path = os.path.join(os.getcwd(),  f"iters_{iters}_simulations", "sims_1samples")
seed = 32778826 # randomly generated seed 
```

```python
results1_df = run_simulations(n_trials,
                             num_mutations,
                             num_cells, 
                             num_CNAs,
                             num_nodes,
                             num_samples,
                             num_regions=num_regions,
                             save_path=save_path,
                             include_LoPhy=True,
                             include_COMPASS=False,
                             include_SCITE=False,
                             include_LACE=False,
                             LoPhy_kwargs={"iters": iters, "het_precision":4.0, "hom_precision":15.0, "fp":0.02, "fn":0.05},
                             seed=seed)
```

```python
results1_df.to_csv(os.path.join(save_path, "sim_evaluation.csv"))
```
